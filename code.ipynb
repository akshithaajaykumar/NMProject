```python
#cnc-regression-continuous
import pandas as pd
from sklearn.linear_model import LinearRegression
path = '/content/drive/MyDrive/Dataset/cncenergy.csv'
df = pd.read_csv(path)
y = pd.to_numeric(df['Operation time'], errors='coerce').fillna(df['Operation time'].mean())
df = df.drop(columns=['energyusage', 'Operation time'])
df_encoded = pd.get_dummies(df)
x = df_encoded.apply(lambda col: pd.to_numeric(col, errors='coerce').fillna(col.mean()))
model = LinearRegression()
model.fit(x, y)
input_row = x.iloc[0].values.reshape(1, -1)
prediction = model.predict(input_row)
print("Prediction:", prediction)
print("Score:", model.score(x, y))

```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/drive/MyDrive/Dataset/cncenergy.csv')
df.columns = df.columns.str.strip()  # Clean column names

# Line Chart: Energy usage over index
plt.figure(figsize=(8, 4))
plt.plot(df['energyusage'], label='Energy Usage', color='green')
plt.title('Energy Usage Over Time')
plt.xlabel('Entry Index')
plt.ylabel('Energy Usage')
plt.grid(True)
plt.legend()
plt.show()

# Bar Chart: Average energy usage by material type
material_avg = df.groupby('Material type')['energyusage'].mean()
material_avg.plot(kind='bar', color='skyblue', figsize=(6,4))
plt.title('Average Energy Usage by Material Type')
plt.ylabel('Average Energy Usage')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()

# Pie Chart: Machine type distribution
machine_counts = df['machine type'].value_counts()
machine_counts.plot(kind='pie', autopct='%1.1f%%', figsize=(5,5), startangle=140)
plt.title('Machine Type Distribution')
plt.ylabel('')
plt.show()
```

```python
#pneumatic-classification-discrete
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
path = '/content/drive/MyDrive/Dataset/pneumatic.csv'
df = pd.read_csv(path)
y = df['Valve_status']
x = df.drop(columns=['Valve_status'])
x = pd.get_dummies(x)
if y.dtype == 'object':
    y = LabelEncoder().fit_transform(y)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
model = KNeighborsClassifier()
model.fit(x_train, y_train)
prediction = model.predict([x.iloc[0].values])
print("Prediction:", prediction)
accuracy = model.score(x_test, y_test)
print("Accuracy:", accuracy)
```

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
path = '/content/drive/MyDrive/Dataset/pneumatic.csv'
df = pd.read_csv(path)
y = df['Valve_status']
x = df.drop(columns=['Valve_status'])
x = pd.get_dummies(x)
if y.dtype == 'object':
    y = LabelEncoder().fit_transform(y)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
model = LogisticRegression(max_iter=1000)
model.fit(x_train, y_train)
prediction = model.predict([x.iloc[0].values])
print("Prediction:", prediction)
accuracy = model.score(x_test, y_test)
print("Accuracy:", accuracy)
```
